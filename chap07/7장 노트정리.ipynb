## 7장 노트정리

#### 모듈이란?  
• 함수,변수 또는 클래스 등을 모아 놓은 파일  
• .py 확장자를 가진 파이썬 파일이 모듈 역할을 함  
• 모듈을 사용하면 코드를 여러 파일로 나누어 작성하여 제사용성과 유지보수성을 높일 수 있습니다.

#### -모듈 사용하기 : import문  
• 다른 파일에 있는 코드를 현재 프로그램으로 불러오는 명령어  
• import 모듈이름 : 모듈 전체를 가져옴. 사용할때는 모듈이름.함수이름() 또는 모듈이름.변수이름 형태로 접근해야 함  
• from 모듈이름 import 함수/변수/클래스 : 모듈에서 특정 항목만 가져옴. 이 경우, 항목을 바로 사용할 수 있습니다  
• import 모듈이름 as 별명 : 모듈에 별명(alias)을 부여하여 코드를 간결하게 만듭니다.
